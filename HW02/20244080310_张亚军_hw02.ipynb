```python
# 1. is_palindrome 判断回文，忽略空格、大小写
def is_palindrome(s: str) -> bool:
    # 去掉空格，全部转为小写
    processed = s.replace(" ", "").lower()
    return processed == processed[::-1]


# 2. is_abecedarian 字母升序，忽略空格
def is_abecedarian(s: str) -> bool:
    processed = s.replace(" ", "").lower()
    for i in range(len(processed)-1):
        if processed[i] > processed[i+1]:
            return False
    return True


# 3. remove_vowels 删除元音 a e i o u
def remove_vowels(s: str) -> str:
    vowels = {"a", "e", "i", "o", "u", "A", "E", "I", "O", "U"}
    res = []
    for ch in s:
        if ch not in vowels:
            res.append(ch)
    return "".join(res)


# --------测试样例--------
print("===is_palindrome测试===")
print(is_palindrome("level"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("abc"))

print("\n===is_abecedarian测试===")
print(is_abecedarian("adder"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("cat"))

print("\n===remove_vowels测试===")
print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
```

```python
#1 list_reverse，输入非列表抛异常
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("参数必须是列表")
    return lst[::-1]


#2 is_sorted：只遍历一次，判断序列升序
def is_sorted(seq):
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True


# 测试
print("===list_reverse测试===")
print(list_reverse([1,2,3]))

print("\n===is_sorted测试===")
print(is_sorted([1,2,3,4]))
print(is_sorted([1,3,2]))
print(is_sorted("abcde"))
```

```python
#1 vec_scalar_mult 向量标量乘法
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not all(isinstance(x, (int, float)) for x in t):
        raise TypeError("元组元素必须为整数或浮点数")
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是整数或浮点数")
    return tuple(x * s for x in t)


#2 vec_inner_product 向量内积
def vec_inner_product(x, y):
    if not (isinstance(x, tuple) and isinstance(y, tuple)):
        raise TypeError("输入必须是元组")
    if len(x) != len(y):
        raise ValueError("两个向量元组长度必须相等")
    if not (all(isinstance(i, (int,float)) for i in x) and all(isinstance(i, (int,float)) for i in y)):
        raise TypeError("向量元素必须是数字")
    return float(sum(a*b for a,b in zip(x,y)))


#3 check_valid_mx 检查有效矩阵：元组的元组
def check_valid_mx(mx):
    #外层必须是元组
    if not isinstance(mx, tuple):
        return False
    #每一行必须是元组
    for row in mx:
        if not isinstance(row, tuple):
            return False
    #所有行长度相同
    row_len = len(mx[0]) if mx else 0
    for row in mx:
        if len(row) != row_len:
            return False
    #每个元素是int/float
    for row in mx:
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True


#4 mx_vec_mult 矩阵向量乘法
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise TypeError("输入不是合法矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    n_col = len(mx[0]) if mx else 0
    if len(vec) != n_col:
        raise ValueError("矩阵列数与向量维度不匹配")
    result = []
    for row in mx:
        result.append(vec_inner_product(row, vec))
    return tuple(result)


# --------测试--------
print("===vec_scalar_mult===")
print(vec_scalar_mult((1,2,3), 2))

print("\n===vec_inner_product===")
print(vec_inner_product((1,2),(3,4)))

print("\n===check_valid_mx===")
test_mx = ((1,2),(3,4))
print(check_valid_mx(test_mx))

print("\n===mx_vec_mult===")
print(mx_vec_mult(test_mx, (1,1)))
```

```python
class Vector:
    """向量类，面向对象实现向量运算"""
    def __init__(self, dim, elements=None):
        # 校验维度
        if not isinstance(dim, int):
            raise TypeError("维度dim必须为整数")
        if dim < 0:
            raise ValueError("维度不能是负数")

        # 只提供维度，不给elements，默认全部填0
        if elements is None:
            elements = [0]*dim

        # 校验元素
        if not all(isinstance(x, (int, float)) for x in elements):
            raise TypeError("向量元素必须是整数或浮点数")
        if len(elements) != dim:
            raise ValueError(f"维度{dim}与元素数量{len(elements)}不一致")

        # 使用实例属性存储维度和元素
        self.dim = dim
        self.elements = tuple(elements)

    """
    Q：维度dim和elements是类属性还是实例属性？
    答案：使用实例属性(self.dim, self.elements)。
    理由：每个Vector实例拥有自己独立的维度与数据；不同向量维度、元素各不相同。
    如果作为类属性，则全部对象共享同一套维度与数据，无法表示多个不同向量。
    """

    # --------比较运算符重载：== < <= > --------
    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的Vector")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的Vector")
        return self.elements < other.elements

    def __le__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的Vector")
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的Vector")
        return self.elements > other.elements

    # dot内积方法
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot参数必须是Vector实例")
        if self.dim != other.dim:
            raise ValueError("向量维度不一致，无法计算内积")
        return sum(a*b for a,b in zip(self.elements, other.elements))

    # *运算符重载：支持标量左/右乘，向量逐元素相乘（哈达玛积）
    def __mul__(self, other):
        # v * 标量
        if isinstance(other, (int, float)):
            new_elems = [x * other for x in self.elements]
            return Vector(self.dim, new_elems)
        # v * w 向量逐元素相乘
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("向量维度不一致，无法逐元素相乘")
            new_elems = [a*b for a,b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        else:
            raise TypeError("不支持该类型乘法")

    # 标量左乘：2 * v
    def __rmul__(self, scalar):
        return self * scalar

    def __repr__(self):
        return f"Vector(dim={self.dim}, elements={list(self.elements)})"


# =========测试Vector类=========
print("===Vector类测试===")
v1 = Vector(2, [2,3])
v2 = Vector(2, [4,5])
print(v1)
print(v2)

print("\nv1 * v2（逐元素相乘）：", v1*v2)
print("2 * v1标量左乘：", 2*v1)
print("v1 *3标量右乘：", v1*3)

print("\nv1.dot(v2)内积：", v1.dot(v2))

v3 = Vector(2, [2,4])
v4 = Vector(2, [2,5])
print("\nv3 < v4 ?", v3 < v4)
print("v3 == v4 ?", v3 == v4)
```
