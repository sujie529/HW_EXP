```python
np.random.seed(42)
scores = np.random.normal(loc=70, scale=10, size=1000)
scores_clip = np.clip(scores,0,100)
scores_nan = scores_clip.copy()
scores_nan[[9,19,29]] = np.nan
valid_score = pd.Series(scores_nan).dropna()

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

# 直方图
plt.figure(figsize=(6,4))
plt.hist(valid_score, bins=20, color="skyblue", edgecolor="black")
plt.title("模拟成绩直方图")
plt.xlabel("分数")
plt.ylabel("频数")
plt.show()

#箱线图
plt.figure(figsize=(5,3))
plt.boxplot(valid_score)
plt.title("成绩箱线图")
plt.ylabel("分数")
plt.show()

# y=x, x², x³ 0~5
x = np.linspace(0,5,100)
y1 = x
y2 = x**2
y3 = x**3
plt.figure(figsize=(7,4))
plt.plot(x,y1,label="$y=x$")
plt.plot(x,y2,label="$y=x^2$")
plt.plot(x,y3,label="$y=x^3$")
plt.legend()
plt.title("函数图像对比")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(alpha=0.3)
plt.show()
```

```python
records = {
    "Name": ["Alice", "Bob", "Carol", "David", "Eva", "Frank"],
    "Group": ["A", "A", "B", "B", "A", "B"],
    "Math": [88, 72, 91, 65, 79, 84],
    "English": [90, 68, 85, 73, 92, 77],
    "Python": [95, 80, 89, 70, 86, 82],
}
df = pd.DataFrame(records)
df = df.set_index("Name") #姓名做行索引

#1 新增Total总分 Average平均分
df["Total"] = df["Math"] + df["English"] + df["Python"]
df["Average"] = df["Total"] / 3

print("行索引：", df.index.tolist())
print("列名：", df.columns.tolist())
print("\n前3行：")
print(df.head(3))
print("\n描述统计：")
print(df.describe())

#2 loc iloc读取Carol；筛选平均分>=80
print("\n==== Carol记录 loc ====")
print(df.loc["Carol"])
print("\n==== Carol记录 iloc（Carol是第2行，下标1）====")
print(df.iloc[1])

high_avg = df[df["Average"] >= 80]
print("\n平均分>=80学生：")
print(high_avg)

#3 axis=0列均值，axis=1行均值
col_mean = df[["Math","English","Python"]].mean(axis=0)
row_mean = df[["Math","English","Python"]].mean(axis=1)
print("\n各科列均值(axis=0):\n", col_mean)
print("\n每个学生行均值(axis=1):\n", row_mean)

#4 apply求单科最高分；agg多统计；rank排名
df["max_subject"] = df[["Math","English","Python"]].apply(lambda s: s.max(), axis=1)
print("\n每名学生单科最高分：")
print(df["max_subject"])

agg_res = df[["Math","English","Python"]].agg(["min","mean","max"])
print("\n每科min mean max agg结果：")
print(agg_res)

df["Math_rank"] = df["Math"].rank(ascending=False)
print("\n数学成绩排名：")
print(df[["Math","Math_rank"]])

#5 groupby分组；transform组内Python平均分
group_obj = df.groupby("Group")
group_count = group_obj.size()
group_avg = group_obj[["Math","English","Python"]].mean()
print("\n每组人数：")
print(group_count)
print("\n每组三科平均分：")
print(group_avg)

df["Group_Python_mean"] = df.groupby("Group")["Python"].transform("mean")
print("\n增加每组Python平均分字段后df：")
print(df)

#6 filter保留组Python平均分>=80
group_filtered = df.groupby("Group").filter(lambda g: g["Python"].mean()>=80)
print("\nfilter过滤后（组Python平均分>=80）")
print(group_filtered)

#7 协方差、皮尔逊相关
subj = df[["Math","English","Python"]]
cov_matrix = subj.cov()
corr_matrix = subj.corr()
print("\n协方差矩阵：")
print(cov_matrix)
print("\n皮尔逊相关系数矩阵：")
print(corr_matrix)

print("""
解读示例：
协方差符号为正：两门科目成绩同向变化，一门高另一门倾向高；负代表反向。
相关系数取值[-1,1]；越接近1强正相关，接近‑1强负相关；接近0几乎无线性相关。
""")
```
