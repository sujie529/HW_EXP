```python
# ========== 1. 热身：简单函数 ==========
def say_hi():
    print("Hello, world!")

def goat_pad(s):
    print("goat" + s + "goat")

def print_n(s, n):
    for _ in range(n):
        print(s)

# 测试
say_hi()
goat_pad("bird")
print_n("hello",3)
```

```python
# ========== 2. 欧几里得算法 GCD ==========
def gcd(m, n):
    a, b = m, n
    while b != 0:
        rem = a % b
        a = b
        b = rem
    return a

# 题目三组测试
print(gcd(2018,2019))
print(gcd(1200,300))
print(gcd(5040,60))
# 负数测试
print(gcd(-24, -18))
```

```python
import math
# ========== 3. 欧拉常数e近似 ==========

def euler_limit(n):
    return (1 + 1/n) ** n

def euler_infinite_sum(n):
    total = 0.0
    fact = 1
    for k in range(n):
        if k > 0:
            fact *= k
        total += 1 / fact
    return total

def euler_approx(epsilon):
    e_true = math.exp(1)
    k = 0
    total = 0.0
    fact = 1
    while True:
        if k>0:
            fact *=k
        total += 1/fact
        if abs(total - e_true) < epsilon:
            return total
        k +=1

def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum = {val}")

#测试
print(euler_limit(10000))
print(euler_infinite_sum(10))
print(euler_approx(0.0001))
print_euler_sum_table(8)
```

```python
# ========== 4. 判断整数幂次 ==========
def is_power_of_2(n):
    if n <=0:
        return False
    while n>1:
        if n%2 !=0:
            return False
        n = n//2
    return True

def is_power(b, n):
    if n <=0 or b <=1:
        return False
    while n % b == 0:
        n = n // b
    return n == 1

#测试
print(is_power_of_2(8))
print(is_power_of_2(10))
print(is_power(3,81))
print(is_power(2,12))
```
