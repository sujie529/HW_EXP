```python
# ========== 单元格1：身份与环境记录 ==========
import hashlib
import os
import platform
import uuid
import sys

# ========= 修改这里为你的学号 =========
STUDENT_ID = 20244080310
# =====================================

CONDA_ENV = os.environ.get("CONDA_DEFAULT_ENV")
print(f"[Conda 环境] {CONDA_ENV}")

MACHINE_FP = hashlib.sha256(
    f"{platform.node()}|{platform.machine()}|{platform.system()}|{uuid.getnode()}".encode()
).hexdigest()[:16]

RUN_ID = f"{STUDENT_ID}-{MACHINE_FP}"
print(f"[学号] {STUDENT_ID}")
print(f"[机器指纹] {MACHINE_FP}")
print(f"[RUN_ID] {RUN_ID}")
print(f"[Python] {sys.version.split()[0]}")
```

```python
# ========== 单元格2：实验1 正则表达式基础匹配 ==========
import re

# 任务1
text1 = "cat on mat; later: cats and dogs"
res_d = re.findall(r"\d+", text1)
res_s = re.findall(r"\s+", text1)
res_bcat = re.findall(r"\bcat\b", text1)
print("=== 任务1 ===")
print(r"\d+ 匹配：", res_d)
print(r"\s+ 匹配：", res_s)
print(r"\bcat\b 匹配：", res_bcat)

# 任务2
print("\n=== 任务2 ===")
text2 = "cat dog bird CAT"
res_catdog = re.findall(r"cat|dog", text2)
print("cat|dog：", res_catdog)

r1 = re.findall(r"a|aa|aaa", "aaaa")
r2 = re.findall(r"aaa|aa|a", "aaaa")
print('re.findall(r"a|aa|aaa", "aaaa") →', r1)
print('re.findall(r"aaa|aa|a", "aaaa") →', r2)
print("说明：| 分支是**顺序优先**，前面匹配成功就不再尝试后面分支")

# 任务3
print("\n=== 任务3 量词测试 ===")
text3 = "ct cat caat caaat"
pats = [
    ("ca*t", r"ca*t  *：0次或多次a"),
    ("ca+t", r"ca+t  +：1次或多次a"),
    ("ca?t", r"ca?t  ?：0次或1次a"),
    ("ca{2}t", r"ca{2}t {2}：恰好2次a"),
    ("ca{1,2}t", r"ca{1,2}t {1,2}：1~2次a")
]
for pat, desc in pats:
    ret = re.findall(pat, text3)
    print(f"{desc} → {ret}")
```

```python
# ========== 单元格3：实验2 分组、回溯引用与替换 ==========
import re

# 任务1：提取邮箱用户名、域名
print("=====任务1 邮箱捕获组=====")
mail_text = "联系:alice@example.com;备选:bob@test.org"
mail_pat = r"([a-zA-Z0-9]+)@([a-zA-Z0-9.]+)"
mails = re.findall(mail_pat, mail_text)
for user, domain in mails:
    print(f"用户名:{user}, 域名:{domain}")

# 任务2：后向引用，找连续重复单词
print("\n=====任务2 连续重复单词=====")
dup_text = "cat cat, dog cat, bird bird"
dup_pat = r"\b(\w+)\b \b\1\b"
dup_res = re.findall(dup_pat, dup_text)
print("重复出现的单词：", dup_res)

# 任务3：日期格式转换 年‑月‑日 → 月‑日‑年，兼容 - / . 分隔
print("\n=====任务3 日期转换=====")
date_text = """2024-05-01
2024/05/02
2024.05.03"""
print("【转换前】")
print(date_text)
# 捕获：年、月、日；分隔符为 - / .
date_pat = r"(\d{4})([-/.])(\d{2})([-/.])(\d{2})"
date_out = re.sub(date_pat, r"\3-\5-\1", date_text)
print("\n【转换后】")
print(date_out)

# 任务4：电话号码统一格式 (123) 456-7890
print("\n=====任务4 电话号码格式标准化=====")
tel_text = "联系我:123-456-7890 或 987.654.3210"
print("原文本：", tel_text)
tel_pat = r"(\d{3})[-.](\d{3})[-.](\d{4})"
tel_out = re.sub(tel_pat, r"(\1) \2-\3", tel_text)
print("格式化后：", tel_out)
```

```python
# ========== 单元格4：实验3 BeautifulSoup解析HTML ==========
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body> <h1>数据分析</h1> <p class="course">第六讲:正则表达式</p> <p class="course">第七讲:结构化数据</p> <a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a> </body></html>
"""

# 任务1：构建解析树，输出标题、h1
print("====任务1====")
soup = BeautifulSoup(html_doc, "html.parser")
title_text = soup.title.get_text()
h1_tag = soup.h1
print("页面标题文本：", title_text)
print("h1标签名：", h1_tag.name)
print("h1文本：", h1_tag.get_text())

# 任务2 find_all 获取class=course的p、所有a链接
print("\n====任务2====")
p_list = soup.find_all("p", class_="course")
print("class='course' 的p段落：")
for p in p_list:
    print("  ", p.get_text())

a_list = soup.find_all("a")
print("\n所有链接：")
for a in a_list:
    link_text = a.get_text()
    href = a["href"]
    print(f"  链接文字：{link_text} , href={href}")

# 任务3 get_text() 对比：整个文档、单个p、单个a
print("\n====任务3 get_text()对比====")
print("整个文档get_text():")
print(repr(soup.get_text()))

print("\n第一个p标签get_text():")
print(repr(p_list[0].get_text()))

print("\n第一个a标签get_text():")
print(repr(a_list[0].get_text()))
```
